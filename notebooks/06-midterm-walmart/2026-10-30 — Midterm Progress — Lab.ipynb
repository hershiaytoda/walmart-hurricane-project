### Q1 — Demand surge

**TODO:** your baseline window, your surge window, the average for each, and the multiple. Plus the chart. State the window boundaries explicitly — a multiple without a window is not a finding.

```python

import pandas as pd
import matplotlib.pyplot as plt

# Make sure timestamps and numeric columns are in the correct format
tx['timestamp'] = pd.to_datetime(
    tx['timestamp'], errors='coerce', format='mixed'
)
tx['quantity'] = pd.to_numeric(tx['quantity'], errors='coerce')

# Remove rows missing information needed for this analysis
tx = tx.dropna(subset=['timestamp', 'quantity', 'category']).copy()

# Standardize category names
tx['category'] = tx['category'].str.strip().str.title()

# Combine inconsistent category labels
category_mapping = {
    'Alcoholic Beverages': 'Alcohol',
    'Baby Products': 'Baby',
    'Breakfast Foods': 'Breakfast',
    'Emergency Supplies': 'Emergency',
    'Games & Entertainment': 'Entertainment',
    'Groceries': 'Grocery',
    'Health & Wellness': 'Health',
    'Household Essentials': 'Household'
}

tx['category'] = tx['category'].replace(category_mapping)

# Create a date column
tx['date'] = tx['timestamp'].dt.normalize()

# Define baseline and surge periods
baseline_start = pd.Timestamp('2024-09-01')
baseline_end = pd.Timestamp('2024-09-15')

surge_start = pd.Timestamp('2024-09-19')
surge_end = pd.Timestamp('2024-09-25')

# Summarize units sold by category and date
daily_category_sales = (
    tx.groupby(['category', 'date'])['quantity']
    .sum()
    .reset_index()
)

# Create a complete category-date combination
all_dates = pd.date_range(baseline_start, surge_end)
categories = daily_category_sales['category'].unique()

complete_index = pd.MultiIndex.from_product(
    [categories, all_dates],
    names=['category', 'date']
)

# Fill dates without sales with zero
daily_category_sales = (
    daily_category_sales
    .set_index(['category', 'date'])
    .reindex(complete_index, fill_value=0)
    .reset_index()
)

# Separate baseline and surge periods
baseline_sales = daily_category_sales[
    (daily_category_sales['date'] >= baseline_start) &
    (daily_category_sales['date'] <= baseline_end)
]

surge_sales = daily_category_sales[
    (daily_category_sales['date'] >= surge_start) &
    (daily_category_sales['date'] <= surge_end)
]

# Calculate average daily units for each category
baseline_avg = (
    baseline_sales.groupby('category')['quantity']
    .mean()
    .rename('baseline_avg_units')
)

surge_avg = (
    surge_sales.groupby('category')['quantity']
    .mean()
    .rename('surge_avg_units')
)

# Combine results
category_comparison = pd.concat(
    [baseline_avg, surge_avg],
    axis=1
).reset_index()

# Calculate absolute and percentage increases
category_comparison['absolute_increase'] = (
    category_comparison['surge_avg_units'] -
    category_comparison['baseline_avg_units']
)

category_comparison['percent_increase'] = (
    category_comparison['absolute_increase'] /
    category_comparison['baseline_avg_units']
) * 100

# Rank categories by percentage increase
category_comparison = category_comparison.sort_values(
    'percent_increase',
    ascending=False
).reset_index(drop=True)

# Display final results
display(category_comparison.round(2))

# Display the top five categories
print("Top 5 categories by relative increase:")
display(category_comparison.head(5).round(2))
```

```python
# TODO: the chart
# Select the top five categories by percentage increase
top_5_categories = category_comparison.head(5)['category'].tolist()

# Filter daily sales for the top five categories
top_5_daily = daily_category_sales[
    daily_category_sales['category'].isin(top_5_categories)
].copy()

# Create the line chart
plt.figure(figsize=(12, 6))

for category in top_5_categories:
    category_data = top_5_daily[
        top_5_daily['category'] == category
    ]

    plt.plot(
        category_data['date'],
        category_data['quantity'],
        marker='o',
        label=category
    )

# Mark the start of the seven-day surge period
plt.axvline(
    pd.Timestamp('2024-09-19'),
    linestyle='--',
    color='red',
    label='Start of Surge Period'
)

plt.title('Daily Unit Sales for Top 5 Categories Before Hurricane Frances')
plt.xlabel('Date')
plt.ylabel('Units Sold')
plt.xticks(rotation=45)
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python

import pandas as pd
import requests
import time

# Open-Meteo historical weather API
BASE = "https://archive-api.open-meteo.com/v1/archive"

# HTTP errors that we should retry
RETRY_STATUSES = {429, 500, 502, 503, 504}


def daily_weather(start, end, lat=28.54, lon=-81.38, tries=3):
    """
    Download daily historical weather data from Open-Meteo.

    Returns:
        A DataFrame with date, precipitation_sum,
        windspeed_10m_max, and source columns.
    """

    params = {
        "latitude": lat,
        "longitude": lon,
        "start_date": start,
        "end_date": end,
        "daily": "precipitation_sum,windspeed_10m_max",
        "timezone": "America/New_York"
    }

    # Retry the request if a temporary error occurs
    for attempt in range(1, tries + 1):
        try:
            response = requests.get(
                BASE,
                params=params,
                timeout=10
            )

        except (requests.Timeout, requests.ConnectionError):
            # Retry connection errors and timeouts
            if attempt == tries:
                raise

            time.sleep(2 ** attempt)
            continue

        # Retry rate limits and temporary server errors
        if response.status_code in RETRY_STATUSES:
            if attempt == tries:
                response.raise_for_status()

            time.sleep(2 ** attempt)
            continue

        # Raise any other unsuccessful HTTP response
        response.raise_for_status()

        # Convert the API response to a DataFrame
        data = pd.DataFrame(response.json()["daily"])

        # Rename time and convert it to datetime immediately
        data = data.rename(columns={"time": "date"})
        data["date"] = pd.to_datetime(data["date"])

        # Identify the source of the weather data
        data["source"] = "open-meteo"

        return data



# Get historical weather for the full month of September 2024
try:
    weather = daily_weather(
        start="2024-09-01",
        end="2024-09-30"
    )

    print("Weather API call successful!")
    print("Number of weather rows:", len(weather))
    print("Source:", weather["source"].iloc[0])
    print("Date column type:", weather["date"].dtype)
    print("Weather timezone: America/New_York")
    display(weather.head())

except requests.RequestException as err:
    print("Weather API call failed:", err)
    print("Check your internet connection and try again.")
```

```python
# Get Orlando's coordinates from the store reference file
orlando_location = stores[stores["store_id"] == "FL-239"].iloc[0]

latitude = orlando_location["latitude"]
longitude = orlando_location["longitude"]

# Retrieve historical weather for September 2024
weather = daily_weather(
    start="2024-09-01",
    end="2024-09-30",
    lat=latitude,
    lon=longitude
)

print("Weather rows:", len(weather))
print("Weather source:", weather["source"].unique())
print(weather.head())

```

```python
# Join each transaction to its corresponding daily weather
sales_weather = orlando.merge(
    weather,
    on="date",
    how="left",
    validate="many_to_one",
    indicator=True
)

# Check that the join did not multiply or remove transaction rows
print("Transaction rows before join:", len(orlando))
print("Transaction rows after join:", len(sales_weather))

print("\nJoin results:")
print(sales_weather["_merge"].value_counts())

# Verify that every original transaction is retained
assert len(sales_weather) == len(orlando)

# Count transactions that successfully matched weather
print("\nTransactions matched to weather:",
      (sales_weather["_merge"] == "both").sum())

print("Transactions without a weather match:",
      (sales_weather["_merge"] == "left_only").sum())

# Aggregate Orlando sales into daily totals
daily_sales = (
    orlando.dropna(subset=["date"])
    .groupby("date")
    .agg(
        daily_revenue=("revenue", "sum"),
        daily_quantity=("quantity", "sum"),
        transaction_count=("transaction_id", "count")
    )
    .reset_index()
)

# Join daily sales to daily weather
daily_analysis = daily_sales.merge(
    weather,
    on="date",
    how="left",
    validate="one_to_one",
    indicator=True
)

print("Daily sales observations:", len(daily_sales))
print("Daily observations after weather join:", len(daily_analysis))
print("\nDaily join results:")
print(daily_analysis["_merge"].value_counts())

# Verify that the daily join preserved all sales dates
assert len(daily_analysis) == len(daily_sales)
```

```python
# Calculate correlations between daily sales and weather variables
correlation_matrix = daily_analysis[
    [
        "daily_revenue",
        "daily_quantity",
        "precipitation_sum",
        "windspeed_10m_max"
    ]
].corr()

print(correlation_matrix.round(3))
import matplotlib.pyplot as plt

# Daily revenue vs. precipitation
plt.figure(figsize=(8, 5))
plt.scatter(
    daily_analysis["precipitation_sum"],
    daily_analysis["daily_revenue"]
)

plt.xlabel("Daily Precipitation (mm)")
plt.ylabel("Daily Revenue ($)")
plt.title("Orlando Daily Revenue vs. Precipitation")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()


# Daily revenue vs. maximum wind speed
plt.figure(figsize=(8, 5))
plt.scatter(
    daily_analysis["windspeed_10m_max"],
    daily_analysis["daily_revenue"]
)

plt.xlabel("Maximum Daily Wind Speed (km/h)")
plt.ylabel("Daily Revenue ($)")
plt.title("Orlando Daily Revenue vs. Maximum Wind Speed")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```
